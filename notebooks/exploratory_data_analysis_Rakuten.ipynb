Analyse exploratoire préliminaire

 Pour assurer la reproductibilité de ton projet, les fichiers sont télécharger avec des chemins dynamiques.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
%matplotlib inline

 # Pour remonter d'un dossier pour le notebook
 # afin d'assurer une meilleure reproductibilité
base_path = os.path.abspath("..") 
data_path = os.path.join(base_path, "data")

X_train = pd.read_csv(os.path.join(data_path, 
                                   "X_train_update.csv"), 
                                   index_col=0)
X_test = pd.read_csv(os.path.join(data_path, 
                                  "X_test_update.csv"), 
                                  index_col=0)
Y_train = pd.read_csv(os.path.join(data_path, 
                                   "Y_train_CVw08PX.csv"), 
                                   index_col=0)
pd.set_option('display.max_colwidth', 25) 
display(X_train.head())
```

```python
modalite_X_train = pd.DataFrame({
    'colonne': X_train.columns,
    'nb_modalites_uniques': [X_train[col].nunique(dropna=False) for col in X_train.columns]
})

modalite_X_train.sort_values(by='nb_modalites_uniques', ascending=False)
```

```python
print("Nombre de produits dans X_train est de:", len(X_train))
```

```python
modalite_X_test = pd.DataFrame({
    'colonne': X_test.columns,
    'nb_modalites_uniques': [X_test[col].nunique(dropna=False) for col in X_test.columns]
})

modalite_X_test.sort_values(by='nb_modalites_uniques', ascending=False)
```

```python
print("Nombre de produits dans X_test est de:", len(X_test))
print("Nombre de produits dans la base Rakuten est de:", len(X_test) + len(X_train))
```

```python
modalite_Y_train = pd.DataFrame({
    'colonne': Y_train.columns,
    'nb_modalites_uniques': [Y_train[col].nunique(dropna=False) for col in Y_train.columns]
})

modalite_Y_train.sort_values(by='nb_modalites_uniques', ascending=False)
```

```python
print("Nombre de catégories de produits dans Y_train est de:", Y_train["prdtypecode"].nunique())
```
